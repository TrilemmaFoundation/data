# ECCC Air Quality Health Index

Real-time Canadian AQHI observations from ECCC for reviewing health-risk index readings at reporting regions and stations.

- Guide: https://data.trilemma.foundation/datasets/eccc-aqhi-observations
- Official source: https://api.weather.gc.ca/collections/aqhi-observations-realtime?f=json
- Data terms: https://open.canada.ca/en/open-government-licence-canada
- Last verified: 2026-09-28

## Overview

ECCC's GeoMet collection exposes preliminary AQHI observations as GeoJSON. Start with five latest features, retaining their observation times and AQHI scale. AQHI is a Canadian health-risk index and is not interchangeable with a U.S. EPA AQI or an individual pollutant value.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Read the official AQHI collection description and Canadian open licence.
2. Request five latest GeoJSON items in one call.
3. Inspect observation time, geometry, and AQHI property.

## Python Example

```python
import requests

response = requests.get(
    "https://api.weather.gc.ca/collections/aqhi-observations-realtime/items",
    params={"f": "json", "latest": "true", "limit": 5}, timeout=30,
)
response.raise_for_status()
data = response.json()
for item in data["features"][:5]:
    print(item["id"], item["properties"].get("aqhi"),
          item["properties"].get("observation_datetime"))
```
