# Met Eireann Weather Warnings

Official Irish weather-warning JSON from Met Éireann for checking county regions, severity, and validity intervals.

- Guide: https://data.trilemma.foundation/datasets/met-eireann-warnings
- Official source: https://www.met.ie/about-us/specialised-services/open-data
- Data terms: https://www.met.ie/about-us/specialised-services/widgets
- Last verified: 2026-09-28

## Overview

Met Éireann exposes current warnings as JSON with CAP IDs, regions, and issue/onset/expiry times. Start with five records. Its reuse terms require attribution and preserving warning meaning; remove expired warnings and do not treat an empty list as a general safety statement.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Read Met Éireann's open-data and weather-warning reuse conditions.
2. Fetch the national JSON warning list once.
3. Retain region IDs and full validity intervals when displaying warnings.

## Python Example

```python
import requests

response = requests.get(
    "https://www.met.ie/Open_Data/json/warning_IRELAND.json", timeout=30,
)
response.raise_for_status()
for warning in response.json()[:5]:
    print(warning["capId"], warning["level"], warning["regions"],
          warning["onset"], warning["expiry"])
```
