# HRSA Area Health Resources Files County Data

HRSA's annually refreshed county health-resource archive for comparing clinician supply and health-system capacity.

- Guide: https://data.trilemma.foundation/datasets/hrsa-ahrf-county
- Official source: https://data.hrsa.gov/data/download?AHRF=&data=AHRF
- Data terms: https://data.hrsa.gov/data/data-sources?tab=DataUsage
- Last verified: 2026-09-28

## Overview

HRSA publishes a county CSV archive and records no usage limitation. Start with one county row and a named clinician variable from the 2024-2025 release. The release year is not every variable's measurement year, and supply does not prove access to care.

## Python Example

```python
import csv
import io
import zipfile
import requests

url = "https://data.hrsa.gov/DataDownload/AHRF/AHRF_2024-2025_CSV.zip"
response = requests.get(url, timeout=60)
response.raise_for_status()
with zipfile.ZipFile(io.BytesIO(response.content)) as archive:
    name = next(n for n in archive.namelist() if n.endswith("AHRF2025.csv"))
    with archive.open(name) as file:
        row = next(csv.DictReader(io.TextIOWrapper(file, encoding="utf-8-sig")))
print(row["fips_st_cnty"], row["phys_nf_prim_care_pc_exc_rsdt_23"])
```
