# SMHI Water Shortage Messages

Official Swedish water-shortage information messages from the impact-based warning API, with dated affected-area and validity metadata for local water-risk review.

- Guide: https://data.trilemma.foundation/datasets/smhi-water-shortage-messages
- Official source: https://opendata-download-warnings.smhi.se/ibww/api/version/1
- Data terms: https://www.smhi.se/data/om-smhis-data/villkor-for-anvandning
- Last verified: 2026-09-28

## Overview

The documented SMHI impact-based warning API also publishes official water-shortage information messages. Start with one current warning list, filter by the documented event code, and keep message text and its timing intact; a message without valid time and area context is not a local warning.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Review the official API entry and the special warning and message terms.
2. Request the current warning JSON once and select WATER_SHORTAGE events.
3. Check affected areas and validity before displaying an unchanged message.

## Python Example

```python
import requests

response = requests.get(
    "https://opendata-download-warnings.smhi.se/ibww/api/version/1/warning.json",
    timeout=20,
)
response.raise_for_status()
messages = [warning for warning in response.json()
            if warning.get("event", {}).get("code") == "WATER_SHORTAGE"]
print(f"{len(messages)} current water-shortage messages")
for message in messages[:3]:
    for area in message.get("warningAreas", [])[:2]:
        name = area.get("areaName", {})
        print(message["id"], name.get("en") or name.get("sv"),
              area.get("published"), area.get("approximateStart"))
```
