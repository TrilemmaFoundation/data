## Python Example

```python
import requests

url = 'https://api.binance.com/api/v3/ticker/price?symbol=BTCUSDT'
response = requests.get(url, timeout=20)
response.raise_for_status()
quote = response.json()
print(quote["symbol"], quote["price"])
```
