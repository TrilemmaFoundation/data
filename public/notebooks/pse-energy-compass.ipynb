# PSE Energy Compass Advisories

Polish grid operator PSE's hourly national electricity-use recommendations for building dated system-context tools.

- Guide: https://data.trilemma.foundation/datasets/pse-energy-compass
- Official source: https://api.raporty.pse.pl/api/pdgsz?$filter=is_active%20eq%20true&$orderby=dtime_utc%20desc&$first=48
- Data terms: https://www.pse.pl/bip/ponowne-wykorzystanie-informacji-publicznej
- Last verified: 2026-09-28

## Overview

PSE publishes the Energy Compass forecast through its official reports API. Start with the first active hourly row, keeping its UTC time and publication time. The advice describes Poland-wide electricity-system conditions and cannot establish a local power outage.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Read PSE's Energy Compass explanation and public-information reuse conditions.
2. Fetch a bounded set of active hourly recommendations.
3. Record the publication time, valid hour, and usage forecast code.

## Python Example

```python
import requests

response = requests.get(
    "https://api.raporty.pse.pl/api/pdgsz",
    params={"$filter": "is_active eq true", "$orderby": "dtime_utc desc", "$first": 48},
    timeout=20,
)
response.raise_for_status()
for row in response.json()["value"][:2]:
    print(row["dtime_utc"], row["usage_fcst"], row["publication_ts_utc"])
```
