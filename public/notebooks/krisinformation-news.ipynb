## Python Example

```python
import requests

url = "https://api.krisinformation.se/v3/news"
response = requests.get(url, params={"format": "json", "allcounties": "true",
                                     "days": 365}, timeout=20)
response.raise_for_status()
for notice in response.json()[:2]:
    print(notice.get("Identifier"), notice.get("Headline"),
          notice.get("Updated"))
```
