# FCDO Foreign Travel Advice

Published UK government country travel advice for building a destination briefing with dated safety context.

- Guide: https://data.trilemma.foundation/datasets/fcdo-travel-advice
- Official source: https://www.gov.uk/api/content/foreign-travel-advice/switzerland
- Data terms: https://www.gov.uk/help/reuse-govuk-content
- Last verified: 2026-09-28

## Overview

The GOV.UK Content API returns structured advice for one country path. Start with Switzerland and inspect its update date; advice is not a guarantee of safety or a substitute for the full official notice.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Read GOV.UK's Content API documentation and reuse terms.
2. Fetch one country's published advice by its GOV.UK path.
3. Inspect the publication date and link to the complete advice.

## Python Example

```python
import requests

url = "https://www.gov.uk/api/content/foreign-travel-advice/switzerland"
response = requests.get(url, timeout=20)
response.raise_for_status()
advice = response.json()
print(advice["title"], advice.get("public_updated_at"))
print("https://www.gov.uk" + advice["base_path"])
```
