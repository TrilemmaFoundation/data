Lithuanian river-station water levels and temperatures for dated local hydrology context tools.

- Guide: https://data.trilemma.foundation/datasets/meteo-lt-hydrology-observations
- Official source: https://api.meteo.lt/
- Data terms: https://api.meteo.lt/
- Last verified: 2026-09-28

## Overview

Meteo LT publishes current measured water levels at named hydro-stations in centimeters. Start with one documented station; a water level without a station datum or official threshold is not a flood warning.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Read the hydro-station endpoint documentation and CC BY-SA terms.
2. Fetch the latest observations for one documented station.
3. Keep its UTC observation time, station name, and water-level unit together.

## Python Example

```python
import requests

response = requests.get(
    "https://api.meteo.lt/v1/hydro-stations/nemajunu-vms/observations/measured/latest",
    timeout=20,
)
response.raise_for_status()
data = response.json()
for observation in data["observations"][-2:]:
    print(data["station"]["name"], observation["observationTimeUtc"],
          observation["waterLevel"], "cm")
```
