## Python Example

```python
import requests

url = 'https://www.bitstamp.net/api/v2/ticker/btcusd/'
response = requests.get(url, timeout=20)
response.raise_for_status()
quote = response.json()
print(quote["timestamp"], quote["last"], quote["volume"])
```
