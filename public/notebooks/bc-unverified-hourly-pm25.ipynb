## Overview

B.C. publishes an unverified PM2.5 CSV and a station metadata CSV. Start with five lines from the current pollutant file using a streaming read; the full file is several megabytes. The catalog record confirms its OGL-BC licence and warns that Metro Vancouver and Fraser Valley readings are excluded. These concentrations are not provider-reported AQI values.

## Python Example

```python
import csv
import itertools
import requests

url = ("https://www.env.gov.bc.ca/epd/bcairquality/aqo/csv/"
       "Hourly_Raw_Air_Data/Air_Quality/PM25.csv")
with requests.get(url, stream=True, timeout=30) as response:
    response.raise_for_status()
    rows = csv.DictReader(response.iter_lines(decode_unicode=True))
    for row in itertools.islice(rows, 5):
        print(row["STATION_NAME"], row["DATE_PST"], row["RAW_VALUE"], row["UNITS"])
```
