# Aviation Weather Center METAR and Stations

Current airport METAR observations and station metadata from NOAA's Aviation Weather Center for checking representative weather conditions near destinations.

- Guide: https://data.trilemma.foundation/datasets/awc-metar-stations
- Official source: https://aviationweather.gov/data/api/
- Data terms: https://sos.noaa.gov/copyright/
- Last verified: 2026-09-28

## Overview

AWC serves airport observations and station metadata through one documented API family. Start with a single airport's METAR and station record. A nearby airport is only a proxy for destination conditions, and this feed does not provide official destination-wide weather warnings.

## Prerequisites

- Python 3.10 or newer

## Python Example

```python
import requests

headers = {"User-Agent": "TrilemmaDataCatalogExample/1.0 (https://data.trilemma.foundation)"}
base = "https://aviationweather.gov/api/data"
params = {"ids": "EIDW", "format": "json"}
metar = requests.get(f"{base}/metar", params=params, headers=headers, timeout=30)
metar.raise_for_status()
station = requests.get(f"{base}/stationinfo", params=params, headers=headers, timeout=30)
station.raise_for_status()
if metar.json() and station.json():
    print(metar.json()[0]["icaoId"], metar.json()[0]["reportTime"],
          station.json()[0]["lat"], station.json()[0]["lon"])
```
