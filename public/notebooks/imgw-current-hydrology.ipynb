# IMGW Poland Current Hydrology

Current Polish river-station observations and official thresholds from IMGW for checking stage exceedance at reviewed gauges.

- Guide: https://data.trilemma.foundation/datasets/imgw-current-hydrology
- Official source: https://danepubliczne.imgw.pl/api/data/hydro/
- Data terms: https://danepubliczne.imgw.pl/
- Last verified: 2026-09-28

## Overview

IMGW's public API returns river-stage measurements with official warning and alarm thresholds. Free access is limited to qualifying noncommercial analysis; commercial and specified sectoral uses can require a separate agreement and payment. Start with five stations from one response. A threshold crossing at one station is not a published area-wide warning.

## Python Example

```python
import requests

response = requests.get("https://danepubliczne.imgw.pl/api/data/hydro/", timeout=30)
response.raise_for_status()
for station in response.json()[:5]:
    print(station.get("id_stacji"), station.get("stan_wody"),
          station.get("stan_ostrzegawczy"), station.get("stan_wody_data_pomiaru"))
```
