## Python Example

```python
import requests

response = requests.get(
    "https://rapidmapping.emergency.copernicus.eu/backend/dashboard-api/public-activations-info/",
    params={"limit": 1, "offset": 0}, timeout=25,
)
response.raise_for_status()
for activation in response.json()["results"]:
    print(activation["code"], activation["name"], activation["eventTime"])
```
