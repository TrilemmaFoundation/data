# Census Building Permits Survey

County counts of new privately owned housing units authorized by building permits for building local housing-supply and market-sizing tools.

- Guide: https://data.trilemma.foundation/datasets/census-building-permits
- Official source: https://www.census.gov/construction/bps/
- Data terms: https://www.usa.gov/government-copyright
- Last verified: 2026-08-20

## Overview

The Building Permits Survey publishes comma-delimited county files for new privately owned residential construction. Start with the latest annual county file and a few rows. Annual files impute nonresponse, and monthly files cover only a sample of permit offices.

## Prerequisites

- Python 3.10 or newer
- A notebook environment such as Jupyter or Google Colab
- An internet connection

## Access the data

1. Open the Building Permits Survey page and confirm you want county authorized-unit counts, not housing starts.
2. Download the documented annual county ASCII file from the Census BPS file directory.
3. Keep FIPS state and county codes, unit counts, and the survey year.

## Python example

```python
import pandas as pd

permits = pd.read_csv(
    "https://www2.census.gov/econ/bps/County/co2025a.txt",
    skiprows=3,
    header=None,
    names=[
        "year",
        "state_fips",
        "county_fips",
        "region",
        "division",
        "county_name",
        "bldgs_1",
        "units_1",
        "value_1",
        "bldgs_2",
        "units_2",
        "value_2",
        "bldgs_34",
        "units_34",
        "value_34",
        "bldgs_5plus",
        "units_5plus",
        "value_5plus",
    ],
    usecols=range(18),
    nrows=30,
    skipinitialspace=True,
)
permits["retrieved_at_utc"] = pd.Timestamp.now(tz="UTC")
print(
    permits[
        [
            "year",
            "state_fips",
            "county_fips",
            "county_name",
            "units_1",
            "units_5plus",
        ]
    ].head(20)
)
```
