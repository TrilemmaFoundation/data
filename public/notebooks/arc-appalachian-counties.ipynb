## Overview

## Python Example

```python
import html
import re
import requests

url = "https://www.arc.gov/appalachian-counties-served-by-arc/"
response = requests.get(url, timeout=20)
response.raise_for_status()
match = re.search(
    r'<strong><a href="https://www.arc.gov/alabama/">Alabama</a>.*?</strong>(.*?)</p>',
    response.text, re.S,
)
if not match:
    raise ValueError("ARC Alabama county list not found")
names = html.unescape(re.sub(r"<[^>]+>", "", match.group(1))).strip()
print(names[:160])
```
