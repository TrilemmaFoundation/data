# Bitview Bitcoin Series

Bitcoin Research Kit (BRK) on-chain time series through Bitview's hosted API for tracking network trends and prototyping bounded research dashboards.

- Guide: https://data.trilemma.foundation/datasets/bitview-bitcoin-series
- Official source: https://bitview.space/api
- Data terms: https://github.com/bitcoinresearchkit/mono/blob/main/LICENSE
- Last verified: 2026-09-28

## Overview

Bitview serves BRK's Bitcoin network series through a free, no-key API. Start with seven daily supply values rather than the full catalog. The supply series measures bitcoin in unspent outputs, not coins available for immediate spending; raw values also need their index to be dated correctly.

## Prerequisites

- Python 3.10 or newer
- An internet connection
- The requests Python package

## Access the Data

1. Open the Bitview API documentation and inspect the Series endpoints.
2. Check the supply series metadata for its definition and day1 index.
3. Request only the last seven day1 values before expanding the range.

## Python Example

```python
import requests

response = requests.get(
    "https://bitview.space/api/series/supply/day1/data",
    params={"start": -7},
    headers={"User-Agent": "Trilemma Data guide (https://data.trilemma.foundation)"},
    timeout=20,
)
response.raise_for_status()
supply_btc = response.json()
print(f"{len(supply_btc)} recent daily supply values (BTC):")
print(supply_btc)
```
