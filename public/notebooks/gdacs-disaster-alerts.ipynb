# GDACS Disaster Alerts

Global multi-hazard disaster alerts for building international emergency monitors that stay linked to official national sources.

- Guide: https://data.trilemma.foundation/datasets/gdacs-disaster-alerts
- Official source: https://www.gdacs.org/
- Data terms: https://www.gdacs.org/
- Last verified: 2026-08-17

## Overview

GDACS publishes multi-hazard alerts as a searchable GeoJSON event list. Start with one hazard type and a recent date window. Alerts are modeled impact estimates for humanitarian awareness; they are not official warnings and must not be used for decisions without national or scientific sources.

## Prerequisites

- Python 3.10 or newer
- A notebook environment such as Jupyter or Google Colab
- An internet connection

## Access the Data

1. Open the GDACS homepage and note the requirement to acknowledge GDACS as the source.
2. Request the SEARCH event list for earthquakes in a bounded recent window.
3. Preserve event type, alert level, country, and retrieval time.

## Python Example

```python
import pandas as pd
import requests

response = requests.get(
    "https://www.gdacs.org/gdacsapi/api/events/geteventlist/SEARCH",
    params={
        "eventlist": "EQ",
        "fromDate": "2026-07-01",
        "toDate": "2026-08-17",
        "alertlevel": "Orange;Red",
    },
    timeout=30,
)
response.raise_for_status()
features = response.json().get("features", [])
alerts = pd.json_normalize(features)
alerts["retrieved_at_utc"] = pd.Timestamp.now(tz="UTC")
print(alerts.head())
```
